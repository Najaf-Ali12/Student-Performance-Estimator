# Student performance Measurer

##  importing required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
from sklearn.preprocessing import RobustScaler
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from xgboost import XGBRegressor
from catboost import CatBoostRegressor
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import r2_score,root_mean_squared_error,mean_squared_error,mean_absolute_error

```

## Loading the data

```python
data=pd.read_csv("Student_Performance_measurement.csv")
data.columns
```

## Data Analysis
There are total 10000 rows in this data
There are 6 columns out of which only one column (Extracurricular Activaties) is empty
data has no missing values

```python
data.info()
data.describe()
```

## Data Encoding

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data['Extracurricular Activities']=le.fit_transform(data['Extracurricular Activities'])
data['Extracurricular Activities'].value_counts()
#It encoded No as 0 and yes as 1
```

## Machine Learning

### Splitting the data

```python
X=data.drop('Performance Index', axis=1)
y=data['Performance Index']

#train test split
X_train,X_test,y_train,y_test=train_test_split(X,y, train_size=0.8, random_state=42)
```

### Selecting the best model
I have tested many models and i found that for this dataset the linear regression is the 
best model. That's why i remove the code of selecting best model and just use linear regression.

```python
model=LinearRegression()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
print("The r2_Score is ",r2_score(y_test,y_pred))
print("The root mean squared error is ",root_mean_squared_error(y_test,y_pred))
print("The mean squared error is ",mean_squared_error(y_test,y_pred))
print("The mean absolute error is ",mean_absolute_error(y_test,y_pred))
print("Cross Validation Score is ",cross_val_score(model,X_train,y_train).mean())
```

### Saving the model

```python
import pickle 
pickle.dump(model,open("Student_Performance_estimator_model.pkl",'wb'))
```
